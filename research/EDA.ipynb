```python
cols = ['speed', 'current_engine_rpm', 'torque']
df[cols].hist(bins=50, figsize=(12,6))
plt.suptitle("Core Telemetry Distributions", fontsize=14)
plt.show()
```

```python
corr = df[['speed','power','torque','gear','steer','acceleration','brake']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm')
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['timestamp_ms'], df['speed'], label='Speed (km/h)')
plt.plot(df['timestamp_ms'], df['gear']*30, label='Gear ×30', alpha=0.7)
plt.legend()
plt.xlabel('Timestamp (ms)')
plt.title('Speed vs Gear Over Time')
plt.show()
```

```python
df['lap_number'].value_counts()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot(df['position_x'], df['position_y'], df['position_z'], linewidth=0.5)
ax.set_title('Track Path (3D)')
ax.set_xlabel('X'); ax.set_ylabel('Y'); ax.set_zlabel('Z')
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['timestamp_ms'], df['acceleration'], label='Throttle', color='limegreen', alpha=0.7)
plt.plot(df['timestamp_ms'], df['brake'], label='Brake', color='red', alpha=0.6)
plt.plot(df['timestamp_ms'], df['steer'], label='Steer', color='blue', alpha=0.6)
plt.legend()
plt.xlabel("Timestamp (ms)")
plt.title("Driver Inputs Over Time")
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='power', y='torque', data=df, alpha=0.4)
plt.title("Power vs Torque Distribution")
plt.xlabel("Power (kW)")
plt.ylabel("Torque (N·m)")
plt.grid(alpha=0.3)
plt.show()
```

```python
tire_cols = [col for col in df.columns if "tire_temp" in col]
plt.figure(figsize=(10,5))
df[tire_cols].plot(kind='box', figsize=(10,5), grid=True)
plt.title("Tire Temperature Distribution per Wheel")
plt.ylabel("Temperature (°C)")
plt.show()
```

```python
slip_cols = [col for col in df.columns if "tire_slip_angle" in col]
plt.figure(figsize=(10,5))
df[slip_cols].plot(kind='line', alpha=0.6)
plt.title("Tire Slip Angle over Time (Front & Rear)")
plt.xlabel("Frame Index")
plt.ylabel("Slip Angle (°)")
plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(x='current_engine_rpm', y='speed', data=df, alpha=0.3, color='purple')
plt.title("RPM vs Speed Correlation")
plt.xlabel("Engine RPM")
plt.ylabel("Speed (km/h)")
plt.grid(alpha=0.3)
plt.show()
```

```python
corr = df.corr(numeric_only=True)
plt.figure(figsize=(12,8))
sns.heatmap(corr[['speed', 'torque', 'power', 'acceleration']].corr(), annot=True, cmap='coolwarm')
plt.title("Correlation among performance metrics")
plt.show()
```

```python
plt.figure(figsize=(10,5))
lap_speed = df.groupby("lap_number")["speed"].mean()
sns.barplot(x=lap_speed.index, y=lap_speed.values, palette="magma")
plt.title("Average Speed per Lap")
plt.xlabel("Lap Number")
plt.ylabel("Average Speed (km/h)")
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    x='position_x', y='position_y', 
    hue=np.where(df['brake']>0, 'Brake', 'Throttle'), 
    palette={'Brake':'red','Throttle':'green'},
    s=3, alpha=0.6, data=df
)
plt.title("Throttle vs Brake Zones on Track Map")
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df['distance_traveled'], df['speed'], color='orange', linewidth=0.8)
plt.title("Speed vs Distance Trace")
plt.xlabel("Distance Traveled (m)")
plt.ylabel("Speed (km/h)")
plt.grid(alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.countplot(x='gear', data=df, palette='viridis')
plt.title("Gear Usage Frequency")
plt.xlabel("Gear")
plt.ylabel("Count")
plt.show()
```

```python
tire_temp_cols = [c for c in df.columns if "tire_temp" in c]
lap_temps = df.groupby("lap_number")[tire_temp_cols].mean()

plt.figure(figsize=(10,6))
sns.heatmap(lap_temps, annot=False, cmap="coolwarm")
plt.title("Average Tire Temperature per Lap")
plt.xlabel("Tires")
plt.ylabel("Lap")
plt.show()
```

```python
motion_cols = [c for c in df.columns if any(x in c for x in ['acceleration','velocity','angular_velocity','suspension','tire_slip'])]
plt.figure(figsize=(12,10))
sns.heatmap(df[motion_cols].corr(), cmap='coolwarm', cbar=False)
plt.title("Correlation Across Motion & Suspension Features")
plt.show()
```

```python
plt.figure(figsize=(14,6))
plt.plot(df['timestamp_ms'], df['speed'], label='Speed (km/h)', alpha=0.7)
plt.plot(df['timestamp_ms'], df['current_engine_rpm'], label='RPM', alpha=0.7)
plt.xlabel('Timestamp (ms)')
plt.ylabel('Values')
plt.title('Speed and RPM over Time')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,8))
plt.scatter(df['position_x'], df['position_y'], c=df['speed'], cmap='viridis', s=1)
plt.colorbar(label='Speed')
plt.title('Car Path Colored by Speed')
plt.xlabel('Position X')
plt.ylabel('Position Y')
plt.show()
```

```python
fig, axs = plt.subplots(3, 1, figsize=(14, 8), sharex=True)

axs[0].plot(df['timestamp_ms'], df['acceleration'], color='g', label='Acceleration')
axs[1].plot(df['timestamp_ms'], df['brake'], color='r', label='Brake')
axs[2].plot(df['timestamp_ms'], df['steer'], color='b', label='Steer Angle')

for ax in axs:
    ax.legend()
    ax.grid(True)
plt.xlabel('Timestamp (ms)')
plt.suptitle('Driver Input Over Time')
plt.show()
```

```python
sns.scatterplot(x='current_lap_time', y='speed', data=df, alpha=0.3)
plt.title('Speed vs Lap Time')
plt.show()
```

```python
df['avg_wheel_speed'] = df[['wheel_rotation_speed_front_left',
                             'wheel_rotation_speed_front_right',
                             'wheel_rotation_speed_rear_left',
                             'wheel_rotation_speed_rear_right']].mean(axis=1)
df['avg_wheel_speed'].hist(bins=50, figsize=(10,5))
plt.title('Average Wheel Speed Distribution')
plt.xlabel('Average Wheel Speed')
plt.ylabel('Frequency')
plt.show()
```
